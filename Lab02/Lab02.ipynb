- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# Task 3: Randomness and reproducibility

# a
np.random.seed(2026)
random_number = np.random.rand()
print("T3a - Random number:", random_number)

# b
normal_numbers = np.random.randn(25)
print("T3b - Standard-normal values:", normal_numbers)

# c
np.random.seed(42)
dice = np.random.randint(1, 7, size=10000)

faces, counts = np.unique(dice, return_counts=True)

print("T3c - Dice faces:", faces)
print("T3c - Frequencies:", counts)

# d
probability_six = np.mean(dice == 6)

print("T3d - Estimated P(six):", probability_six)
print("T3d - Theoretical P(six):", 1/6)

# e
two_sixes = np.count_nonzero(
    (dice[:-1] == 6) & (dice[1:] == 6)
)

print("T3e - Consecutive pairs of sixes:", two_sixes)
```

## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).

- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# Task 7: Climate of five cities

# a
print("T7a - Shape of tmax:", tmax.shape)

# b
annual_means = tmax.mean(axis=1)
print("T7b - Cities:", cities)
print("T7b - Annual mean temperatures (C):",
      np.round(annual_means, 2))

# c
hottest_indices = np.argmax(tmax, axis=1)
hottest_months = months[hottest_indices]

print("T7c - Hottest month for each city:", hottest_months)

# d
monthly_means = tmax.mean(axis=0)
print("T7d - Months:", months)
print("T7d - Five-city monthly averages (C):",
      np.round(monthly_means, 2))

# e
hot_month_counts = np.sum(tmax > 35, axis=1)
print("T7e - Number of months above 35 C:",
      hot_month_counts)

# f
lahore_fahrenheit = tmax[0] * 9/5 + 32
print("T7f - Lahore temperatures (F):",
      np.round(lahore_fahrenheit, 2))

# g
monsoon_rain = rain[:, 6:9].sum(axis=1)
print("T7g - Monsoon rainfall (mm):",
      np.round(monsoon_rain, 2))

# h
wettest_index = np.argmax(monsoon_rain)
annual_rain = rain.sum(axis=1)
monsoon_percentage = monsoon_rain / annual_rain * 100

print("T7h - Wettest monsoon city:", cities[wettest_index])
print("T7h - Monsoon rainfall share (%):",
      np.round(monsoon_percentage, 2))

print("\nT7 Interpretation: I was surprised that Islamabad receives more monsoon rainfall than Lahore in the provided climate data. Islamabad receives about 763 mm from July to September, compared with about 454.3 mm in Lahore. I also noticed that Quetta receives only about 5.96% of its annual rainfall during these months, indicating a very different seasonal rainfall pattern.")
```

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# Task 8: Inflation FY2024 vs FY2025

# (a) 
average_inflation = cpi.mean(axis=1)

print("T8a - FY2024 average inflation:",
      round(average_inflation[0], 2), "%")
print("T8a - FY2025 average inflation:",
      round(average_inflation[1], 2), "%")

# (b) 
high_inflation_months = cpi_months[cpi[0] > 25]

print("T8b - FY2024 months above 25%:",
      high_inflation_months)

# (c) 
inflation_drop = cpi[0] - cpi[1]

print("T8c - Months:", cpi_months)
print("T8c - Inflation drop (percentage points):",
      np.round(inflation_drop, 2))

# (d) 
biggest_drop_index = np.argmax(inflation_drop)

print("T8d - Biggest drop month:",
      cpi_months[biggest_drop_index])
print("T8d - Biggest drop:",
      round(inflation_drop[biggest_drop_index], 2),
      "percentage points")

# (e) 
inflation_labels = np.where(
    cpi[1] < 5,
    "Low",
    np.where(cpi[1] < 10, "Moderate", "High")
)

print("T8e - Months:", cpi_months)
print("T8e - FY2025 categories:", inflation_labels)

print("\nT8 Interpretation: The average year-on-year inflation rate in the provided data decreased from 26.22% in FY2024 to 4.87% in FY2025. January showed the largest year-on-year rate difference between the two years, at 25.9 percentage points. This indicates that prices were generally increasing at a much slower rate in FY2025; it does not necessarily mean that prices fell.")
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# Task 9: Grace marks, pass rate and Z-scores

# (a) Add 5 grace marks and cap at 100
after_grace = np.clip(marks + 5, 0, 100)
print("T9a - Marks after grace:", after_grace)

# (b) Pass percentage (passing marks >= 50)
pass_percentage = np.mean(after_grace >= 50) * 100
print("T9b - Pass percentage:", round(pass_percentage, 2), "%")

# (c) Students who passed only because of grace marks
new_pass_indices = np.where(
    (marks < 50) & (after_grace >= 50)
)[0]

print("T9c - Newly passing student indices:", new_pass_indices)
print("T9c - Student numbers:", new_pass_indices + 1)

# (d) Z-scores of marks after grace
mean_marks = after_grace.mean()
std_marks = after_grace.std()

z_scores = (after_grace - mean_marks) / std_marks

print("T9d - Z-scores:", np.round(z_scores, 3))
```

- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).

- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
# Task 10: An image as a NumPy array

# (a) Create an 8x8 black image
img = np.zeros((8, 8), dtype=int)

# Draw the number 7 using slicing
img[1, 1:7] = 255
img[2:7, 5] = 255
img[4, 3:7] = 128

print("T10a - Original image:\n", img)

# (b) Flip image left to right
flipped = img[:, ::-1]
print("T10b - Flipped image:\n", flipped)

# (c) Invert the colors
inverted = 255 - img
print("T10c - Inverted image:\n", inverted)

# (d) Crop rows 1-4 and columns 3-6
cropped = img[1:5, 3:7]
print("T10d - Cropped image:\n", cropped)

# (e) Brighten all pixels by 80, capped at 255
brightened = np.clip(img + 80, 0, 255)
print("T10e - Brightened image:\n", brightened)

# (f) Shape, dtype and number of white pixels
print("T10f - Image shape:", img.shape)
print("T10f - Data type:", img.dtype)
print("T10f - Number of white pixels:",
      np.count_nonzero(img == 255))
```
